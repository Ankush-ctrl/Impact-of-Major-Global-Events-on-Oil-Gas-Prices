# correcting columns and misspeling

```python
data.head()
```

```python
data = data.rename(columns={'Unnamed: 0' : 'Year'})
```

```python
data = data.drop(0)
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data = data.fillna(method='ffill')
```

# extracting usefull columns

making new data
* oil_data
* naturalgas_data
* GoldSilver_data

```python
data.columns
```

@ Oil Data

```python
oil_data = data[['Year', 'Crude oil, Brent', 'Crude oil, Dubai', 'Crude oil, WTI', 'Coconut oil', 'Groundnut oil', 'Palm oil', 'Palm kernel oil', 'Soybean oil']]
```

```python
oil_data['Year'] = oil_data['Year'].astype(float).astype(int)
```

```python
cols_to_fix = ['Crude oil, Brent', 'Crude oil, Dubai', 'Crude oil, WTI', 'Coconut oil', 'Groundnut oil', 'Palm oil', 'Palm kernel oil', 'Soybean oil']

oil_data[cols_to_fix] = oil_data[cols_to_fix].apply(pd.to_numeric, errors='coerce')
```

```python
print(oil_data[['Year'] + cols_to_fix].dtypes)
```

```python
oil_data.head()
```

@ Natural Gas Data

```python
naturalgas_data = data[['Year', 'Natural gas, US', 'Natural gas, Europe', 'Liquefied natural gas, Japan']]
```

```python
naturalgas_data['Year'] = naturalgas_data['Year'].astype(float).astype(int)
```

```python
cols_to_fix = ['Natural gas, US', 'Natural gas, Europe', 'Liquefied natural gas, Japan']
naturalgas_data[cols_to_fix] = naturalgas_data[cols_to_fix].apply(pd.to_numeric, errors = 'coerce')
```

```python
naturalgas_data.info()
```

@ Metals Data

```python
GoldSilver_data = data[['Year', 'Gold', 'Silver']]
```

```python
GoldSilver_data['Year'] = GoldSilver_data['Year'].astype(float).astype(int)
```

```python
cols_to_fix = ['Gold', 'Silver']
GoldSilver_data[cols_to_fix] = GoldSilver_data[cols_to_fix].apply(pd.to_numeric, errors = 'coerce')
```

```python
GoldSilver_data.info()
```

# Brent VS WTI Crude Oil

```python
oil_data.columns
```

```python
oil_data = oil_data.sort_values('Year')

plt.plot(oil_data['Year'], oil_data['Crude oil, Brent'], label='Brent')
plt.plot(oil_data['Year'], oil_data['Crude oil, WTI'], label='WTI')

plt.title("Brent VS WTI Oil Price Trends")
plt.xlabel("Year")
plt.ylabel("Prices(USD $)")
plt.legend()

plt.grid(True)
plt.show()
```

# Brent VS Dubai Crude Oil

```python
sns.set_style("whitegrid")
sns.lineplot(data=oil_data, x="Year", y="Crude oil, Brent", label="Brent", color = 'royalblue')
sns.lineplot(data=oil_data, x="Year", y="Crude oil, Dubai", label="Dubai", color = 'darkorange')

plt.title("Brent VS Dubai Oil Price Trends")
plt.xlabel("Year")
plt.ylabel("Price (USD $)")
plt.show()
```

# 2008 Financial Crisis impact on Oil 

```python
oil_2004_2011 = oil_data[(oil_data["Year"] >= 2004) & (oil_data["Year"] <= 2011)]
```

```python
sns.barplot(data=oil_2004_2011, x = 'Year', y = 'Crude oil, Brent', palette= 'dark')
plt.title("2008 Financial Crisis")
plt.xlabel("Year")
plt.ylabel("Price (USD $)")
```
